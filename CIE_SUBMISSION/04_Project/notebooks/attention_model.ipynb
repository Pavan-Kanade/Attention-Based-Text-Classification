### 1. Additive Attention Layer Definition
$$\text{score}_i = \tanh(W_h h_i + b)$$
$$\alpha_i = \text{softmax}(\text{score}_i)$$
$$c = \sum \alpha_i h_i$$

```python
class AttentionLayer(nn.Module):
    def __init__(self, hidden_dim):
        super(AttentionLayer, self).__init__()
        self.w_h = nn.Linear(hidden_dim, hidden_dim)
        self.v_a = nn.Linear(hidden_dim, 1, bias=False)

    def forward(self, lstm_outputs):
        u_i = torch.tanh(self.w_h(lstm_outputs))
        scores = self.v_a(u_i).squeeze(-1)
        alpha = F.softmax(scores, dim=-1)
        context = torch.bmm(alpha.unsqueeze(1), lstm_outputs).squeeze(1)
        return context, alpha
```
